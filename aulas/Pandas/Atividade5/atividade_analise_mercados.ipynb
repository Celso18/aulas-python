```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("vendas_mercados - vendas_mercados.csv")
df.head()
```

```python
df["Lucro"] = (
    df["Preco_Unitario"] - df["Custo_Unitario"]
) * df["Quantidade"]

lucro_por_mercado = (
    df.groupby("Mercado")["Lucro"]
      .sum()
      .sort_values(ascending=False)
)
```

```python
df["Lucro"] = (
    df["Preco_Unitario"] - df["Custo_Unitario"]
) * df["Quantidade"]

lucro_por_mercado = (
    df.groupby("Mercado")["Lucro"]
      .sum()
      .sort_values(ascending=False)
)
lucro_por_mercado.round(2)
```

```python
pagamentos_por_regiao = (
    df.groupby(["Regiao", "Forma_Pagamento"])
      .size()
      .reset_index(name="Quantidade")
)

pagamentos_por_regiao
```

```python
quantidade_por_produto = (
    df.groupby("Produto")["Quantidade"]
      .sum()
      .sort_values(ascending=False)
)

quantidade_por_produto
```

```python
media_precos = (
    df.groupby("Mercado")["Preco_Unitario"]
      .mean()
      .sort_values(ascending=False)
)

media_precos.round(2)
```

```python
df["Faturamento"] = (
    df["Preco_Unitario"] * df["Quantidade"]
)

faturamento_por_mercado = (
    df.groupby("Mercado")["Faturamento"]
      .sum()
      .sort_values(ascending=False)
)

print(faturamento_por_mercado.round(2))
print("\nFaturamento total da rede:")
print(round(df["Faturamento"].sum(), 2))
```

```python
df["Preco_Promocional"] = np.where(
    df["Quantidade"] >= 10,
    df["Preco_Unitario"] * 0.90,
    df["Preco_Unitario"]
)

df["Faturamento_Promocional"] = (
    df["Preco_Promocional"] * df["Quantidade"]
)

novo_faturamento = df["Faturamento_Promocional"].sum()

print("Novo faturamento total da rede:")
print(round(novo_faturamento, 2))
```
