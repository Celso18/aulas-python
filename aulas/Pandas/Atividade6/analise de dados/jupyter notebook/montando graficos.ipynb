```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ==========================================
# 1. GRÁFICO DE LINHA SIMPLES (X E Y)
# ==========================================
x = np.array([1, 2, 3, 4, 5])  # eixo horizontal
y = np.array([2, 4, 3, 6, 5])  # eixo vertical

plt.figure(figsize=(8, 4))
plt.plot(x, y, color='red', linestyle='-', marker='o')
plt.title('GRÁFICO DE X E Y', fontsize=20)
plt.xlabel('Horizontal', fontsize=12)
plt.ylabel('Vertical', fontsize=12)
# Adicionando textos explicativos nos pontos máximos e mínimos
plt.text(4, 6, "TOPO", fontsize=15, color='#FF8000')
plt.text(1, 2, "BASE", fontsize=15, color='#FF8000')
plt.grid(True, linestyle=':')
plt.show()

# ==========================================
# 2. PROCESSAMENTO DE DADOS COM PANDAS
# ==========================================
dados_turma = [
    {'nome': 'João', 'idade': 31, 'materia': "Python", 'nota_final': 10},
    {'nome': 'Mateus', 'idade': 19, 'materia': "Python", 'nota_final': 9},
    {'nome': 'Bruno', 'idade': 22, 'materia': "Java", 'nota_final': 5},
    {'nome': 'Chalie', 'idade': 33, 'materia': "Java", 'nota_final': 9},
    {'nome': 'Marcelo', 'idade': 44, 'materia': "Analise de Dados", 'nota_final': 8},
    {'nome': 'Victor', 'idade': 21, 'materia': "Analise de Dados", 'nota_final': 9},
]

df_turmas = pd.DataFrame(dados_turma)

# Agrupando por matéria e tirando a média das notas
media_turmas = df_turmas.groupby(['materia'])["nota_final"].mean()
print("--- Médias por Matéria Calculadas pelo Pandas ---")
print(media_turmas)
print("-" * 50)

# Dados extraídos do agrupamento para alimentar os gráficos
categorias = ["ADS", "JAVA", "PYTH"]
valores = [media_turmas["Analise de Dados"], media_turmas["Java"], media_turmas["Python"]]

# ==========================================
# 3. GRÁFICO DE BARRAS VERTICAIS
# ==========================================
plt.figure(figsize=(6, 4))
plt.bar(categorias, valores, color='royalblue')
plt.title("GRAFICO DE MÉDIAS (VERTICAL)")
plt.xlabel("TURMAS")
plt.ylabel("NOTA FINAL")
plt.grid(True, linestyle=':')
plt.show()

# ==========================================
# 4. GRÁFICO DE BARRAS HORIZONTAIS (CORRIGIDO)
# ==========================================
plt.figure(figsize=(6, 4))
plt.barh(categorias, valores, color="red")
plt.title("GRAFICO DE MÉDIAS (HORIZONTAL)")
plt.xlabel("NOTA FINAL")  # CORREÇÃO: O valor numérico agora fica na horizontal
plt.ylabel("TURMAS")      # CORREÇÃO: As categorias ficam na vertical
plt.grid(True, linestyle=':')
plt.show()

# ==========================================
# 5. GRÁFICO DE LINHA (MÉDIAS)
# ==========================================
plt.figure(figsize=(6, 4))
plt.plot(categorias, valores, color="red", marker='o', linewidth=2)
plt.title("GRAFICO DE MÉDIAS (LINHA)")
plt.xlabel("TURMAS")
plt.ylabel("NOTA FINAL")
plt.grid(True, linestyle=':')
plt.show()

# ==========================================
# 6. GRÁFICO DE ONDA SENOIDE (ESTILO ORIENTADO A OBJETOS)
# ==========================================
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(t, s, color='tab:blue')
ax.set(xlabel='time (s)', ylabel='voltage (mV)', title='GRAFICO SENO')
ax.grid(True, linestyle='--')
fig.savefig("test.png")  # Salva o gráfico atual como imagem na sua pasta
plt.show()
```
